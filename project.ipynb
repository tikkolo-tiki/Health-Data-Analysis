```python

import pandas as pd
import numpy as np


import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', 120)

# Plot style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

```python
# Number of duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='Region',
    order=df['Region'].value_counts().index
)

plt.title("Distribution of Patients by Region")
plt.xlabel("Region")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y='County',
    order=df['County'].value_counts().index
)

plt.title("Distribution of Patients by County")
plt.xlabel("Number of Patients")
plt.ylabel("County")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Year',
    order=sorted(df['Year'].dropna().unique())
)

plt.title("Number of Patient Records by Year")
plt.xlabel("Year")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
month_order = [
    'January', 'February', 'March',
    'April', 'May', 'June',
    'July', 'August', 'September',
    'October', 'November', 'December'
]

plt.figure(figsize=(12, 5))

sns.countplot(
    data=df,
    x='Month',
    order=[
        month for month in month_order
        if month in df['Month'].unique()
    ]
)

plt.title("Patient Records by Month")
plt.xlabel("Month")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
if 'Heart_Rate' in df.columns:

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x='Gender',
        y='Heart_Rate'
    )

    plt.title("Heart Rate Distribution by Gender")
    plt.xlabel("Gender")
    plt.ylabel("Heart Rate")

    plt.tight_layout()
    plt.show()
```

```python
if 'Heart_Rate' in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x='Age',
        y='Heart_Rate',
        hue='Gender'
    )

    plt.title("Age vs Heart Rate")
    plt.xlabel("Age")
    plt.ylabel("Heart Rate")

    plt.tight_layout()
    plt.show()
```

```python
daily_records = (
    df.groupby('Date')
      .size()
      .reset_index(name='Number_of_Records')
)

plt.figure(figsize=(14, 5))

sns.lineplot(
    data=daily_records,
    x='Date',
    y='Number_of_Records'
)

plt.title("Patient Records Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
